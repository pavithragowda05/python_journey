```python
#program to keep accepting numbers and appending to a list
#until user type 0, once 0 is typed then display the list and sum of
#elements in the list.

mylist=[]
a=1
tot=0
while a!=0: #while a>0:
    a = int(input("Enter a number :"))
    mylist.append(a)
    tot+=a
print(mylist)
print(tot)
```

```python
# 1. to generate fibonicci series for n terms using while loop
n=int(input("enter a number:"))
a=0
b=1
print(a)
print(b)
while(a<=n):
    c=a+b
    print(c)
    a=b
    b=c
```

```python
# 2. to find the given number is perfect number or not using while loop
n=int(input("enter a number:"))
a=1
tot=0
while(a<n):
    if n%a==0:
        tot+=a
    a+=1    
        
if tot==n:
    print("perfect number:")
else:
    print("not perfect")
```

```python
# 1. to find the sum of digits in the given number:
#     ex. 456 = 15
n=int(input("enter a number:"))
sum=0
while(n!=0):
    y=n%10
    sum+=y
    n=n//10
print(sum)   
```
