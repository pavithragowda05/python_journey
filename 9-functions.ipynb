```python
#Functions :

# Function is a block of code/set of instructions used to do one specific task, 
# that can be executed multiple times from different parts of our program. 

# Advantages of functions in Python:

# 1. Code Reusability: 
# Functions enable code reusability, reducing duplication and making your 
# code more efficient.

# 2. Modularity: Functions break down large programs into smaller, 
#     manageable modules, making it easier to maintain and update code.

# 3. Readability: Functions improve code readability by providing a 
# clear and concise way to express complex operations.

# 4. Easier Debugging: Functions make it easier to identify and debug 
# errors, as each function has a specific responsibility.

# 5. Improved Collaboration: Functions facilitate collaboration among 
# developers, as each function can be worked on independently.

# 6. Reduced Errors: Functions reduce errors by providing a single point 
# of definition for a specific operation, reducing the likelihood 
# of inconsistent implementations.

# 7. Faster Execution: Functions can improve execution speed by 
# allowing Python to optimize and cache function calls.

# There are two types of functions :
#     1. user defined functions :
#     2. library functions :
```

```python
#function to accept gender as m/f and return Male/Female:

def fn_gender(gen):
    if gen=="m" or gen=="M":
        g="Male"
    elif gen=="f" or gen=="F":
        g="Female"
    else:
        g="Others"
    return g
#implemetation
ge = input("Enter gender :")
ogen = fn_gender(ge)
print(ogen)    
```

```python
#function with list as input param and simple data as return value :
#task : function to accept a list and return biggest element in the list.

def fn_big_element(ilist): #input param for the function is list
    big=0
    for x in ilist:
        if x>big:
            big=x
    return big
```

```python
#implementation :
mylist = [10,20,1,2,3,4,100,200,12,14,15,16]
bignum = fn_big_element(mylist)
print(f"biggest element in the list : {bignum}")
```

```python
#program to accept a simple data and returns a list:
#task : function to accept a number and return multiplication table for
#the given number and products will in list.

def fn_mul_table(n):
    tlist = []
    for x in range(1,11):
        tlist.append(n*x)
    return tlist #return type is a list
```

```python
num = int(input("Enter a number :"))
mlist = fn_mul_table(num)
print(mlist)
```

```python
#function to accept to list objects and return a dictionary
#assumption : elements of first list are keys and element of second list are
#values.

def fn_dict(list1,list2):
    mydict = {}
    for x in list(zip(list1,list2)):
        mydict[x[0]]=x[1]
    return mydict
    list1 = [101,102,103,104]
list2 = ['danush','usha','preksha 2','somashekar']
newdict = fn_dict(list1,list2)
print(newdict)
```

```python
#program to accept a dictionary and return a list of tuples

def fn_dict_listoftuple(mydict):
    dlist = []
    for x in mydict:
        dlist.append((x,mydict[x]))
    return dlist
#implementation
studict = {101: 'danush', 102: 'usha', 103: 'preksha 2', 104: 'somashekar'}
newdict = fn_dict_listoftuple(studict)
print(newdict)
    
```
