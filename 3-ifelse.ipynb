```python
#statements/main statements and sub statements :

all python main statements should be column 1

if  a statement is having substatements then the main statement should
end with : (colon), then the subsequent sub statements will apper in
next indent position.
(indent : one tab/four spaces)

a
b
c:
    x
    y
    z:
        i
        j
        k
d
e
f

a,b,c,d,e,f are main statements
x,y,z are sub statements of c
i,j,k are sub statements of z
```

```python
#conditional statements / decision making statements :

if condition:
    set of instructions

if evaluates the given condition, if the condition is true/stisfied then
it will execute the given set of instructions in its subsequent indent.

if/else:
    
if condition:
    set of instructions
else:
    set of instructions
    
if evaluates the given condition, if the condition is true/stisfied then
it will execute the given set of instructions in its subsequent indent,
otherwise it will execute the set of instructions after else:
```

```python

#if:

amt = int(input("Enter amount :"))
if amt>=10000:
    print("Too costly")
if amt<10000:
    print("Good price")

#if/else :

amt = int(input("Enter amount :"))
if amt>=10000:
    print("Too costly")
else:
    print("Good price")

```

```python
#logical operators :

and : conjunction : checks if all the given condtions are ture
or : disjucntion : check if any one of the given condition is true
not : negation : makes the meaning of the given expresstion to negative
        good
        not good = bad

#using logical operators :

age = int(input("Enter age :"))
if age>=60:
    print("Senior citizen")
if age>=40 and age<60:
    print("Middle age")
if age>=20 and age<40:
    print("Young age")
if age>=13 and age<20:
    print("Teen age")
if age<13:
    print("Child")
```

```python
#if/elif/else : (elif : else if)

if condition:
    set of instructions
elif condition:
    set of instructions
.
.
.
.
.
.
else:
    set of instructions
    
if evaluates the given condition, if the condition is true/stisfied then
it will execute the given set of instructions in its subsequent indent,
otherwise it will evaluate the next elif condition, if the condition is
ture/satisfied then it will execute the given set of instructions in its
subsequent indent,if all the condtions fails then it will execute the set 
of instructions after else:
```

```python
#if/elif/else:

age = int(input("Enter age :"))
if age>=60:
    print("Senior citizen")
elif age>=40 and age<60:
    print("Middle age")
elif age>=20 and age<40:
    print("Young age")
elif age>=13 and age<20:
    print("Teen age")
else:
    print("Child")
    
#note : if/elif/else : is better to use when we are checking same variable
#with different conditions.
```

```python
#membership operators :

#in, not in, any : these operators works with collection of data.

#variable in collection

alist = [20,30,50,60,70]
x = int(input("Enter number to find :"))

if x in alist:
    print("Number is there in list")
else:
    print("Number is not there in list")
```
