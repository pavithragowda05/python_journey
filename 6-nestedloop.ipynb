```python
#nested loops: loop with in a loop

for x loop:
    for y loop:
        for z loop:
            set of instructions for z
        set of instructions for y
    set of instructions for x
    
#in nested loops inner most loop will be completed first.

x loop will start first then enters to y loop, y loop will start
and enters to z, z will start and completes the loop, the controls back
y, after y controls back x
```

```python
#program to generate multiplication tables from 1 to 10

for x in range(1,11):
    for y in range(1,11):
        print(x*y,end = " ")
    print()
```
