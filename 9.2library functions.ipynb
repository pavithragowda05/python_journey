```python
#4. dict1 = {101:1000,102:2000,103:3000}
#dict2 = {101:(200,'cr'),102:(500,'dr')....}

#sample result : dict3 = {101:1200,102:1500...}

dict1 = {101:1000,102:2000,103:3000}
dict2 = {101:(1000,'cr'),102:(500,'dr'),103 :(1000,'dr')}
dict3 = {}
for x in dict1:
    val = dict1[x]
    tup = dict2[x]
    if tup[1]=='cr':
        bal = val + tup[0]
    else:
        bal = val - tup[0]
    dict3[x] = bal
print(dict3)
```

```python
# 2. empname = [10 names] 
# Gender = [] 
# Designation = [manager/hr/admin/technical] 
# Salary = [25000……..] 
# Emplist = [] create a list of tuples using the given data 
# Find the net salary based on the given: 
# HRA 10% of the salary 
# CCA 5% of the salary 
# Grosssalary = Salary + hra + cca 
# Incometax 8% of the salary 
# Netsalary = grosssalary - incometax 
# Final output example: (rama,m,manager,25000,2500,1250,28750,2000,26750)


empname=["pavi","ani","ashu","rithu","abhi","anju","diya"]
gender=["f","m","f","f","m","f","f"]
Designation =["manager","hr","admin","technical","hr","admin","technical"]
salary = [25000,20000,15000,10000,20000,15000,10000] 
hra=[]
cca=[]
grosalary=[]
tax=[]
net=[]
for x in salary:
    h=int(x*(10/100))
    hra.append(h)
    c=int(x*(5/100))
    cca.append(c)
    grosal=x+h+c
    grosalary.append(grosal)
    t=int(x*(8/100))
    tax.append(t)
    netsal=grosal-t
    net.append(netsal)
    
flist = list(zip(empname, gender, Designation, salary, hra, cca, grosalary, tax, net))
for x in flist:
    print(x)

```
